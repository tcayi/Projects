```python
# 雙層迴圈
# 你一定要先固定一個東西
# 九九乘法 n1 * n2 = ?
# 我先固定 n1
# 之前叫你名字用i, 如果是第二層用j
i = 0
while i < 9:
    n1 = i + 1
    # 針對這n1做九次
    j = 0
    while j < 9:
        n2 = j + 1
        result = n1 * n2
        print(n1, "*", n2, "=", result)
        j = j + 1
    i = i + 1
```

```python
import urllib.request as req
import bs4 as bs

url = "https://tabelog.com/tw/tokyo/rstLst/ramen/?SrtT=rt"
resp = req.urlopen(url)
content = resp.read()

html=bs.BeautifulSoup(content)

#Find all of 地點類型，評分，晚餐價錢，午餐價錢，哪天沒開

sect = html.find_all("div",{"class":"list-rst__rst-data"})

for r in sect:

    #r.find這邊記得要對應到相應的tag，有的henf都沒有

    #地點
    loc = r.find("div", {"class":"list-rst__area-genre"})
    loc_text = loc.text
    print(loc_text)


   # 評分
    rating = r.find("span",{"class":"list-rst__rating-val"})
    rating_text = rating.text
    print(rating_text)

    #晚餐價錢
    sup=r.find("span",{"class":"c-rating-v3__val"})
    sup_text = sup.text
    print(sup_text)

    #午餐價錢
    lunch=r.find("span",{"class":"c-rating-v3__val"})
    lunch_text=lunch.text
    print(lunch_text)

    # #哪天沒開
    offday=r.find("span",{"class":"list-rst__holiday-text"})
    offday_text=offday.text
    print(offday_text)





```
